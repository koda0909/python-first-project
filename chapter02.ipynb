### 논리 연산자
- and
- or
- not

````python
is_adult = True
has_ticket = False

# and 연산
print(is_adult and has_ticket) # False
# or 연산
print(is_adult or has_ticket) # True
# not 연산
print(not is_adult) # False
````

### 문자열

````python
print("선발 출전: %s, %d" % ("손흥민", 20))
print("선발 출전: [%5d]" % (20))
print("소숫점 표기 : %.4f" %3.141592)
````

````python
name = "Cake"
price = 6500
qty = 2

# format 방식
format_str = "주문내역: {0} {1} -> 총 {2}원".format(name, qty, price)
print(format_str)

# f-sting 방식
format_str = f"주문내역: {name} {qty} -> 총 {(price * qty):>10,}원"
print(format_str)
````

````python
print(f"금액: {15000:>8,}원")
````

````python
# 가상의 주문 데이터
orders = [
    ("Latte", 5000, 2),
    ("Americano", 3000, 1),
    ("Cake", 6000, 1)
]


print("=" * 36)
print(f"{'메뉴명':<10}{'단가':>8}{'금액':>9}")
print("-" * 36)

# 총금액
total_amount = 0

for menu, price, qty in orders:
    subtotal = price * qty
    total_amount += subtotal
    print(f"{menu:<10} {price:>8} {subtotal:>9}")

print("_" * 36) 
print(f"{'합계금액:':<10}{total_amount:>20}원")   



print (total_amount)
````
